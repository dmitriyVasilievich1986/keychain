## Import libraries

```python
from dotenv import load_dotenv
from pathlib import Path
import json
```

```python
from keychain.commands.backup_db import BackupDBCommand, RestoreDBCommand
from keychain.commands.backup_db.utils import find_backup_files
from keychain.config import AppConfig
```

## Load `.env` files

```python
load_dotenv("../../../../.env")
```

```python
load_dotenv("../../../../db.env")
```

## Initialize constants

```python
app_config = AppConfig.get_or_create()
```

```python
app_config.db.backup_folder_path = Path("../../../../../backup")
```

## Execute main code

#### Find dump files

```python
backup_files = find_backup_files(app_config.db.backup_folder_path, app_config.db.provider, app_config.db.name)
```

```python
print(json.dumps([x.model_dump() for x in backup_files], indent=4, default=str))
```

#### Backup command

```python
cmd = BackupDBCommand(app_config)
```

```python
await cmd.validate()
```

```python
await cmd.execute()
```

#### Restore command

```python
cmd = RestoreDBCommand(app_config)
```

```python
await cmd.validate()
```

```python
await cmd.execute()
```
