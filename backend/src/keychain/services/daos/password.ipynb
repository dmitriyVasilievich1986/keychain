## Import libraries

```python
from dotenv import load_dotenv
```

```python
from keychain.config import AppConfig
from keychain.services.db_client import DBClient
from keychain.services.daos.password import PasswordDAO
from keychain.services.cryptography.client import CryptographyClient
```

```python
from keychain.modules.routers.models.response.password import PasswordGetResponseModel
from keychain.modules.routers.models.request.password import PasswordCreateRequestModel, PasswordUpdateRequestModel
```

## Load `.env` files

```python
load_dotenv("../../../../.env")
```

```python
load_dotenv("../../../../db.env")
```

## Initialize constants

```python
app_config = AppConfig.get_or_create()
```

```python
db_client = DBClient(app_config)
```

```python
cryptography_client = CryptographyClient(app_config)
```

```python
password_dao = PasswordDAO(db_client)
```

## Main code execution

#### Get object

```python
await db_client.healthcheck()
```

```python
password = await password_dao.get_by_pk(1)
```

```python
password_model = PasswordGetResponseModel.model_validate(password)
```

```python
print(password_model.model_dump_json(indent=4))
```

#### Create object

```python
create_body = PasswordCreateRequestModel(
    name="test",
    image_url="/image.svg"
)
await password_dao.create(**create_body.model_dump(), user_id=1)
```

#### Update object

```python
update_body = PasswordUpdateRequestModel(
    name="test",
    image_url="/image.svg"
)
await password_dao.update(**update_body.model_dump())
```

#### Delete object

```python
await password_dao.delete(1)
```
