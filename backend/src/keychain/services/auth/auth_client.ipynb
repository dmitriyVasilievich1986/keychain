## Import libraries

```python
from dotenv import load_dotenv
from datetime import datetime, timedelta
```

```python
from keychain.config import AppConfig
from keychain.services.auth.client import AuthClient
```

## Load `.env` files

```python
load_dotenv("../../../../.env")
```

```python
load_dotenv("../../../../db.env")
```

## Main code execution

```python
app_config = AppConfig.get_or_create()
```

```python
auth_client = AuthClient(app_config)
```

```python
auth_client
```

#### Encode token

```python
expiration_time = datetime.now() + timedelta(minutes=10)
```

```python
auth_token = auth_client.encode_token(sub="example", exp=expiration_time)
```

```python
print(auth_token.model_dump_json(indent=4))
```

#### Decode token

```python
token = auth_token.access_token
# token = "token"
```

```python
auth_client.decode_token(token)
```
